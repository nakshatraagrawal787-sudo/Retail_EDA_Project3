 # OUTLIER ANALYSIS

 ## Outliers are observations that are unusually far from the general distribution of the data.Boxplots and the Interquartile Range (IQR) method are used to identify potential outliers.

 # Boxplot — Product Price

```python
plt.figure(figsize=(8,4))

sns.boxplot(x=data['product_price'])

plt.title('Boxplot of Product Price')
plt.xlabel('Product Price')
plt.show()
```

 # Boxplot — Final Price

```python
plt.figure(figsize=(8,4))

sns.boxplot(x=data['final_price'])

plt.title('Boxplot of Final Price')
plt.xlabel('Final Price')
plt.show()
```

 # IQR Code

```python
Q1 = data['final_price'].quantile(0.25)
Q3 = data['final_price'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data['final_price'] < lower_bound) |
    (data['final_price'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

### Insight

The IQR method identifies potential high-value outliers in 
final_price. These observations should not automatically be 
removed because they may represent genuine high-value 
transactions rather than data errors.

 # Z-SCORE

```python
from scipy.stats import zscore

data['final_price_zscore'] = zscore(data['final_price'])

zscore_outliers = data[
    abs(data['final_price_zscore']) > 3
]

print("Z-score outliers:", len(zscore_outliers))
```

```python
data.drop(columns='final_price_zscore', inplace=True)
```

### Observation

The Z-score method provides another way of identifying unusually 
high or low transaction values. These observations are treated 
as potential outliers and are retained because they may represent 
legitimate retail transactions

#  Skewness Analysis

Skewness measures the asymmetry of a distribution.

- Positive skewness → longer right tail
- Negative skewness → longer left tail
- Near zero → approximately symmetric distribution

```python
skewness = data[numeric_cols].skew()

skewness
```

### Insight

The final_price variable has positive skewness of approximately 
0.935, indicating a right-skewed distribution. This means that 
most transactions are concentrated at lower-to-medium values, 
while a smaller number of high-value transactions extend the 
right tail of the distribution.

 # DISCOUNT ANALYSIS

```python
data['discount_percentage'].dtype
```

```python
data['discount_percentage'] = pd.to_numeric(
    data['discount_percentage'],
    errors='coerce'
)
```

```python
bins = [0, 5, 10, 15, 20, 25, 30]

labels = [
    '0-5%',
    '6-10%',
    '11-15%',
    '16-20%',
    '21-25%',
    '26-30%'
]

data['discount_band'] = pd.cut(
    data['discount_percentage'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
data[['discount_percentage', 'discount_band']].head(20)
```

```python
data['discount_band'].value_counts().sort_index()
```

```python
discount_analysis = data.groupby(
    'discount_band',
    observed=True
).agg(
    transactions=('order_id', 'count'),
    average_final_price=('final_price', 'mean'),
    total_revenue=('final_price', 'sum')
)

discount_analysis
```

```python
discount_analysis['average_final_price'].plot(
    kind='bar',
    figsize=(9,5)
)

plt.title('Average Final Price by Discount Band')
plt.xlabel('Discount Band')
plt.ylabel('Average Final Price')
plt.xticks(rotation=0)
plt.show()
```

### Insight

Lower discount levels are associated with higher average 
transaction values in this dataset. This is consistent with 
the negative correlation between discount_percentage and 
final_price
